```python
import osmnx as ox
ruta_mapa = "../data/mapa_cdmx.graphml" 
G = ox.load_graphml(ruta_mapa)
```

```python
import sys
sys.path.append('../src')
from fase2 import algo_estrella, algo_greedy, heuristica_Haversine

nodo_origen = 3639971455
nodo_destino = 274196207

res_a = algo_estrella(G, nodo_origen, nodo_destino, heuristica_Haversine)
print(f"A*     -> Distancia: {res_a['metros']:.2f} m | Nodos expandidos: {res_a['nodos_expandidos']}")

res_greedy = algo_greedy(G, nodo_origen, nodo_destino, heuristica_Haversine)
print(f"Greedy -> Distancia: {res_greedy['metros']:.2f} m | Nodos expandidos: {res_greedy['nodos_expandidos']}")
```

```python
import folium
nodo_origen = 3639971455
nodo_destino = 274196207
mapa = folium.Map(location=[19.452, -99.189], zoom_start=13,tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}', attr='Esri')

coords_a = []
for nodo in res_a['camino']:
    lat = G.nodes[nodo]['y']
    lon = G.nodes[nodo]['x']
    coords_a.append((lat, lon))
#A*
folium.PolyLine(coords_a, color='green', weight=5).add_to(mapa)

coords_greedy = []
for nodo in res_greedy['camino']:
    lat = G.nodes[nodo]['y']
    lon = G.nodes[nodo]['x']
    coords_greedy.append((lat, lon))
#Greedy
folium.PolyLine(coords_greedy, color='blue', weight=5).add_to(mapa)

origen_coord = [G.nodes[nodo_origen]['y'], G.nodes[nodo_origen]['x']]
destino_coord = [G.nodes[nodo_destino]['y'], G.nodes[nodo_destino]['x']]

folium.Marker(origen_coord, popup="Inicio").add_to(mapa)
folium.Marker(destino_coord, popup="Fin").add_to(mapa)

#mapa
mapa.save("mapa.html")
mapa
```
